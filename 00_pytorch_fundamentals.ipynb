## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - Returna view of an input tensor of certain sahpe but keep the **same memory** as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack or h stack)
* Squeeze - removes all `1` deminsions from a tensor
* Unsqueeze - add a `1` dimension to a target tensors
* Permute - Returna view of the input witih  dimensions permuted (swapped) in a certain way

```python
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1, 9)
z, z.shape
```

```python
# Changin z changes x (becuase the view of hte tensor shares the same memor yas the original input)
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stack = torch.stack([x, x, x, x], dim=1)
x_stacked, x_stack
```

```python
# torch.squeeze()
print(f" Previous tensor: {x_reshaped}")
print(f" Previous shape: {x_reshaped.shape}")

x_squeezed = x_reshaped.squeeze()
print(f"\n New tensor: {x_squeezed}")
print(f" New shape: {x_squeezed.shape}")
```

```python
# torch.squeeze()
print(f" Previous tensor: {x_squeezed}")
print(f" Previous shape: {x_squeezed.shape}")

x_unsqueeze = x_squeezed.unsqueeze(dim=1)
# Add an extra dimension
print(f"\n New tensor: {x_unsqueeze}")
print(f" New shape: {x_unsqueeze.shape}")
```

```python
# torch.permute - rearrange the dimension of tensor i na specified order
x_original = torch.rand(size=(224, 224, 3)) # [height, width, color_channels]

# Permute the original tensor to bring colors in front
x_permuted = x_original.permute(2, 0, 1) # shifts
print(f" The original shape is {x_original.shape}")
print(f" The permuted shape is {x_permuted.shape}")
```

```python
x_original[0, 0, 0,] = 1234
x_original[0, 0, 0], x_permuted[0, 0, 0]
```

## PyTorch tensors and NumPy

* Data in NumPy,-> PyTorch -> `torch.from_numpy(ndarray)
* PyTorch -> NymPy-> `torch.Tensor.numpy()

```python
# NumPy array to tensor
import numpy as np
import torch

array = np.arange(1.0, 8.0) # defualt numpy dtype is float64
tensor = torch.from_numpy(array)
array, tensor
```

```python
 # change the value of the array, what happens? to the `tensor`
array = array + 1
array, tensor
```

```python
# Tensor to NumPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
# change the tensor what happens to numpy_tensor
tensor = tensor + 1
tensor, numpy_tensor
```

## Reproducibility

```python
import torch

# create two random tensors

random_tensor_A = torch.rand(3,4)
random_tensor_B =torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# make them reproducible
import torch

# set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)
print(random_tensor_C == random_tensor_D)
```
